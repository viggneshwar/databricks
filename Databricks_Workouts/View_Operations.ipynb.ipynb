Local temp views exist only within the current Spark session.

```python
# Run a SQL query on the Spark session to select all records from the table 'orders_dv'
df1 = spark.sql("SELECT * FROM lakehousecat1.deltadb.orders_dv")

# Create or replace a temporary view named 'localview1' using the DataFrame 'df1'
# This allows you to query 'df1' later using SQL syntax
df1.createOrReplaceTempView("localview1")

# Run a SQL query on the temporary view 'localview1' to select all records
# Display only the first 2 rows of the result for quick inspection
spark.sql("SELECT * FROM localview1").show(2)
```

Global temp views are shared across notebooks in the same cluster.

```python
# Create or replace a temporary view named 'localview1' from the DataFrame 'df1'
# Temporary views are session-scoped, meaning they are only accessible within the current Spark session.
df1.createOrReplaceTempView("localview1")

# Create or replace a global temporary view named 'globalview1' from the DataFrame 'df1'
# Global temporary views are tied to a system-preserved database called 'global_temp'.
# They can be accessed across multiple Spark sessions using the syntax: SELECT * FROM global_temp.globalview1
df1.createOrReplaceGlobalTempView("globalview1")
```
