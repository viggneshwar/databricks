```python
# Create a text widget named "catalog_new" with default value "default"
# This will show up in the Databricks notebook UI as a text box
dbutils.widgets.text("catalog_new", "default")

# Retrieve the value entered in the "catalog_new" widget
# .strip() removes any leading/trailing spaces from the input
catalog = dbutils.widgets.get("catalog_new").strip()

# Create another text widget named "schema_new" with default value "default"
dbutils.widgets.text("schema_new", "default")

# Retrieve the value entered in the "schema_new" widget
# Again, .strip() ensures clean input without extra spaces
schema = dbutils.widgets.get("schema_new").strip()
# Print a confirmation message to the notebook output
print("Input text objects created")
print(f"Catalog: {catalog}, Schema: {schema}")
```

###Creating catalog and schema

```python
# Print a message showing which catalog is being created
print(f"Creating '{catalog}' catalog")

# Create the catalog if it does not already exist
spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")

# Print a message showing which schema is being created inside the catalog
print(f"Creating '{schema}' schema")

# Create the schema inside the specified catalog if it does not already exist
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
```

###Creating silver,bronze,gold and datasource volumes

```python
# Print a message and create the "gold" volume inside the given catalog and schema
print(f"Creating gold volume in {catalog}.{schema}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog}.{schema}.gold")

# Print a message and create the "silver" volume
print(f"Creating silver volume in {catalog}.{schema}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog}.{schema}.silver")

# Print a message and create the "bronze" volume
print(f"Creating bronze volume in {catalog}.{schema}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog}.{schema}.bronze")

# Print a message and create the "sourcedata" volume
print(f"Creating sourcedata volume in {catalog}.{schema}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog}.{schema}.sourcedata")
```

###Defining variables for paths and databases

```python
# Define the path to the gold volume and print it
gold_path = f"/Volumes/{catalog}/{schema}/gold"
print(f"The gold volume is located at {gold_path}")

# Define the path to the silver volume and print it
silver_path = f"/Volumes/{catalog}/{schema}/silver"
print(f"The silver volume is located at {silver_path}")

# Define the path to the bronze volume and print it
bronze_path = f"/Volumes/{catalog}/{schema}/bronze"
print(f"The bronze volume is located at {bronze_path}")

# Define the path to the sourcedata volume and print it
sourcedata_path = f"/Volumes/{catalog}/{schema}/sourcedata"
print(f"The sourcedata volume is located at {sourcedata_path}")

# Define the silver database reference and print it
silver_db = f"{catalog}.{schema}"
print(f"The silver database is {silver_db}")

# Define the gold database reference and print it
gold_db = f"{catalog}.{schema}"
print(f"The gold database is {gold_db}")
```
